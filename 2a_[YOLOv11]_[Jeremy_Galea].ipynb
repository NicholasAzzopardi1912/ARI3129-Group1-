## YOLOv11

### Imports

```python
from __future__ import annotations

import torch
import json 

from pathlib import Path
from ultralytics import YOLO
```

### Try to use GPU

```python
device = "cuda:0" if torch.cuda.is_available() else "cpu"

print("=== Device check ===")
print("torch.cuda.is_available():", torch.cuda.is_available())
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
    print("CUDA version (torch):", torch.version.cuda)
print("Using device:", device)
print()
```

### Train YOLOv11

```python
model = YOLO("yolo11s.pt")
model.train(
    data="data\\YOLO_COCO\\data.yaml",  # path to the dataset yaml file
    epochs=500,                         # more than enough epochs to train
    batch=8,
    imgsz=640,
    device = device,                    # use GPU if available
    project="runs/YOLO/detect/runs/train",
    name="yolov11_experiment",          # name of the experiment
    exist_ok=True,
    patience=30,                        # early stopping patience (30 epochs)
)
```

### Evaluate the Model on Test Set

#### Validation

```python
eval_model = YOLO("runs/YOLO/detect/runs/train/yolov11_experiment/weights/best.pt")
metrics = eval_model.val(
    data = "data\YOLO_COCO\data.yaml",
    split = "test",
    imgsz = 640,
    conf = 0.25,
    save = True,
    plots = True,
)

print("Precision (P):   ", metrics.box.mp)
print("Recall (R):      ", metrics.box.mr)
print("mAP@0.5:         ", metrics.box.map50)
print("mAP@0.5:0.95:    ", metrics.box.map)

```

#### Predictions:

```python
results = eval_model.predict(
    source="data\\YOLO_COCO\\images\\test",
    imgsz=640,
    conf=0.25,
    save=True,
    plots=True,
)
```
