```python
# Importing YOLOv8 from Ultralytics along with the required utilities
from ultralytics import YOLO
from pathlib import Path
import json
import csv
```

```python
# MODEL TRAINING
# Path to the YOLO dataset configuration
DATA_YAML = r"C:\Users\nicho\OneDrive\University\Year 3\CV\Project\ARI3129 - Assignment Materials\ARI3129 - Assignment Materials\Data\Datasets\YOLO_COCO_sign_shape\data.yaml"

# Loading the pretrained YOLOv8 model
model = YOLO("yolov8s.pt")

# Training the model using the sign shape dataset
model.train(
    data=DATA_YAML,
    epochs=100,
    imgsz=640,
    batch=-1,
    device=0,
    workers=4,
    patience=30,
    project="runs/ari3129",
    name="yolov8_sign_shape_v1"
)
```

```python
# MODEL EVALUATION ON TEST SET
# Path to the best model obtained during training
BEST = r"runs\ari3129\yolov8_sign_shape_v1\weights\best.pt"
# Loading the best model
m = YOLO(BEST)
# Evaluating the performance on the test split
m.val(data=DATA_YAML, split="test", device=0, project="runs/ari3129", name="yolov8_sign_shape_test")
```

```python
# CREATING THE ANALYTICS SETUP TO INTERPRET THE RESULTS
# Weights of the final model
WEIGHTS = r"runs/ari3129/yolov8_sign_shape_v1/weights/best.pt"
# Directory containing the set of test images
IMAGE_PATH = r"C:\Users\nicho\OneDrive\University\Year 3\CV\Project\ARI3129 - Assignment Materials\ARI3129 - Assignment Materials\Data\Datasets\YOLO_COCO_sign_shape\images\test"

# Confidence threshold and IoU threshold
CONF = 0.5
IOU  = 0.50

# Output directory for predictions and analytics
OUT_DIR = Path("runs/ari3129/analytics_sign_shape")
OUT_DIR.mkdir(parents=True, exist_ok=True)

# loading the trained model and mapping the class ID's to readable names
model = YOLO(WEIGHTS)
class_names = model.names

# Function to obtain analytics for each image based on number of detected sign shapes and computing the total number of detected signs
def analytics_for_image(result):
    # Initialising the count for each class
    counts = {name: 0 for _, name in class_names.items()}

    # If the detections exist, count the occurences of each class
    if result.boxes is not None and len(result.boxes) > 0:
        cls_ids = result.boxes.cls.cpu().tolist()
        for cid in cls_ids:
            counts[class_names[int(cid)]] += 1

    total = sum(counts.values())
    return {"counts": counts, "total": total}

# Storing the predictions using the test set of images into a folder
results = model.predict(
    source=IMAGE_PATH,
    conf=CONF,
    iou=IOU,
    device=0,
    save=True,
    project=str(OUT_DIR),
    name="predictions",
    verbose=False
)

# Aggregating the results
analytics = []
for r in results:
    # Extracting the image filename
    img_name = Path(r.path).name if r.path else "unknown"

    # Computing the analytics for the current image
    a = analytics_for_image(r)
    analytics.append({
        "image": img_name,
        "total_signs_detected": a["total"],
        "counts_per_shape": a["counts"]
    })

# Saving the results to a JSON
json_path = OUT_DIR / "sign_shape_analytics.json"
with json_path.open("w", encoding="utf-8") as f:
    json.dump(analytics, f, indent=2)

# Saving the results which includes the image name, total count, and one column per class to a CSV
csv_path = OUT_DIR / "sign_shape_analytics.csv"
fieldnames = ["image", "total_signs_detected"] + [class_names[i] for i in sorted(class_names.keys())]

with csv_path.open("w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    for row in analytics:
        # Flattening nested counts dictionary for CSV format
        flat = {"image": row["image"], "total_signs_detected": row["total_signs_detected"]}
        flat.update(row["counts_per_shape"])
        writer.writerow(flat)

# Printing to the console
print("Saved analytics to:")
print(" -", json_path)
print(" -", csv_path)
print("\n Example output (first 3 images):")

for row in analytics[:3]:
    print(f"\nImage: {row['image']}")
    print("Total signs detected:", row["total_signs_detected"])
    for k, v in row["counts_per_shape"].items():
        if v > 0:
            print(f"  - {k}: {v}")

print("\n Annotated predictions saved under:")
print(OUT_DIR / "predictions")
```
