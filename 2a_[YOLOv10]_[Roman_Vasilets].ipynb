```python
import yaml
from pathlib import Path

ROOT_DIR = Path(r'C:\RemoteStorage\UoM\Year 3\ARI3129 Advanced CV for AI\Object Detection Models')
DATASET_DIR = ROOT_DIR / 'Datasets' / 'YOLO_COCO'
YAML_PATH = DATASET_DIR / 'data.yaml'

train_path = 'images/train'
val_path = 'images/val'
test_path = 'images/test'

data_config = {
    'path': str(DATASET_DIR), 
    'train': train_path, 
    'val': val_path,     
    'test': test_path,   
    'nc': 6,
    'names': [
        'Blind-Spot Mirror (Convex)', 
        'No Entry (One Way)', 
        'No Through Road (T-Sign)', 
        'Pedestrian Crossing', 
        'Roundabout Ahead', 
        'Stop'
    ]
}

with open(YAML_PATH, 'w') as f:
    yaml.dump(data_config, f, default_flow_style=False)
```

```python
from ultralytics import YOLO

model = YOLO('yolov10n.pt') 

results = model.train(
    data=str(YAML_PATH),
    epochs=50,
    imgsz=640,
    batch=16,
    name='YOLO_SignType_v10',
    device='cpu' 
)
```

```python
run_folder = 'YOLO_SignType_v103'
ROOT_DIR = Path(r'C:\RemoteStorage\UoM\Year 3\ARI3129 Advanced CV for AI\Object Detection Models')
best_weights = ROOT_DIR / 'runs' / 'detect' / run_folder / 'weights' / 'best.pt'

model = YOLO(best_weights)

metrics = model.val(split='test') 

print(f"mAP @ 50% IoU:  {metrics.box.map50:.4f}")
print(f"mAP @ 50-95% IoU:     {metrics.box.map:.4f}")
print(f"Precision (Mean):            {metrics.box.mp:.4f}")
print(f"Recall (Mean):               {metrics.box.mr:.4f}")

print(metrics.box.maps)
```
